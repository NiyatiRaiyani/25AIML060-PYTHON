```python
#4.Using Finally Block
try:
    number=int(input("Enter a number:"))
    result=10/number
except ZeroDivisionError:
    print("Division by zero is not allowed.")
except ValueError:
    print("Invalid input! Please enter a numeric value.")
else:
    print("The result of division is:", result)
finally:
    print("Execution completed.")
```

```python
#5.Handling Multiple Errors in One Block
try:
    num1=int(input("Enter the value of num1:"))
    num2=int(input("Enter the value of num2:"))
    print("Result:",num1/num2)
except (ZeroDivisionError, ValueError) as e:
    print("Error occurred:", e)
```

```python
# 6. Create a custom Exception
class underageError(Exception):
    pass
age = int(input("Enter yor Age :"))
try:
    if age < 18:
        raise underageError("You must be age")
    else:
        print("Access grated.")
except underageError as e:
    print("Custom Exception :" ,e)
```

```python
# 7. TypeError Handling in function
def add_numbers(x,y):
    if not isinstance(x,(int,float)) or not isinstance(y,(int,float)):
        raise TypeError("Both Arguments must be numbers.")
    return x+y
try:
    print(add_numbers("10",5))
except TypeError as e:
    print("Error :",e)
```

```python
#8.Using else with try-except
try:
    num=int(input("Enter an even number:"))
    if num%2!=0:
        raise ValueError("The number is not even.")
except ValueError as ve:
    print("Error:", ve)
else:
    print("Thank you for entering an even number:", num)
```

```python
#9. Nested try blocks
try:
    a = int(input("Enter a number :"))
    try:
        b = int(input("Enter another number :"))
        print("Result :" ,a/b)
    except ZeroDivisionError:
        print("Can not Divide by zero.")
except ValueError:
    print("Invalid inputs . Only integers are allowed")
```

```python
#10. Re-raising Exception
def divide(a,b):
    try:
        return a/b
    except ZeroDivisionError as e:
        raise
try:
    divide(10,0)
except ZeroDivisionError:
    print("Handled the re-raised exception in the outer block.")
```

**Question 1
Create a function that will decide if the provided numbers constitute a Pythagorean triplet.**

```python
def pythagorean(a, b, c):
    nums = sorted([a, b, c])
    
    if nums[0]*2 + nums[1]2 == nums[2]*2:
        return True
    else:
        return False

print(pythagorean(3, 4, 5))
print(pythagorean(5, 6, 7))
```

**Question 2
Create a function that will take a list of any length from the user and return the last value in the list.**

```python
def last_value(lst):
    return lst[-1]

numbers = [10, 20, 30, 40, 50]

print(last_value(numbers))
```
