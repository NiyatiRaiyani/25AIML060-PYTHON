```python
#ZIP
seq1 = ["a","b","c"]
seq2 = [2,3,4]

print(list(zip(seq1,seq2)))
```

```python
zipped = zip(seq1,seq2)
for i,(a,b) in enumerate(zipped):
    print(i,a,b)
```
