```python
import numpy as np
arr1 = np.array([[1,2,3],[4,5,6]])
print(arr1)

arr2 = np.array([[7,8,9],[10,11,12]])
print(arr2)
```

```python
arr1+arr2
```

```python
arr3 = np.vstack((arr1,arr2))
print(arr3)

print(arr3.shape)
```

**Horizontal stack**

```python
import numpy as np
arr3 = np.hstack((arr1,arr2))
print(arr3)

print(arr3.shape)
```
