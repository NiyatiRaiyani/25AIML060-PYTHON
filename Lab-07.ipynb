Branching Programs 

Iteration

```python
x = 3
ans = 0
itersleft = x
while(itersleft!=0):
    ans = ans + x
    itersleft = itersleft - 1
print(str(x) + '*' + str(x) + '=' + str(ans))
```

For loops

```python
sequence = [1,2,None,4,None,5]
total = 0
for value in sequence:
    if value is None:
        continue
    total += value
    print(total)
```
