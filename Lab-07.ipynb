1) Find the sum of all integers given In a number, E.g 478, total = 19

```python
# Sum of digits

num = int(input("Enter a number: "))

total = 0
temp = num

while temp > 0:
    digit = temp % 10
    total = total + digit
    temp = temp // 10

print("Sum of digits =", total)
```

2) Reverse the given number, Input = 8965, Output= 5698

```python
# Reverse a number

num = int(input("Enter a number: "))

reverse = 0
temp = num

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp = temp // 10

print("Reversed number =", reverse)
```
