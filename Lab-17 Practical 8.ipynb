112. Write a program to compare loop-based vs vectorized operations using time
measurement
• Implement element-wise multiplication using both loop and vectorized methods.
• Use %timeit or time to benchmark performance.
• Identify performance boost achieved via vectorization.

```python
import numpy as np
import time

a = np.arange(1000000)
b = np.arange(1000000)

start = time.time()

result = np.zeros(1000000)

for i in range(len(a)):
    result[i] = a[i] * b[i]

loop_time = time.time() - start

start = time.time()

result = a * b

vector_time = time.time() - start

print("Loop time:", loop_time)
print("Vectorized time:", vector_time)
```
