**Nested List Comprehension**

```python
matrix = [[i * j for j in range(1, 4)] for i in range(1, 4)]

print(matrix)
```

```python
students = [
    ["Niyati", 20],
    ["Priya", 21],
    ["Krina", 19]
]

result = [[name, data] for name, data in students]

print(result)
```

```python
names = ["Niyati", "Priya", "Neha", "Riya", "Meera",
         "Karan", "Deep", "Krina", "Het", "Pooja"]

result = [name for name in names if 'e' in name.lower()]

print(result)
```

```python
result = [(name, name.lower().count('e')) for name in names]

print(result)
```

```python
name_data=[['John','Emily','Micheal','engineer'],['elephant','javier','natalia']]
```

```python
names_2e=[]
for names in name_data:
    two_es =[name for  name in names if name.count('e')>=2]
    names_2e.extend(two_es)
names_2e
```

**Positional / Keyword Argument**

```python
def a_function(a, b, c=20):
    if c > 1:
        return c * (a + b)
    else:
        return c / (a + b)
```

```python
a_function(8,9,c=5)
```

**Arbitary argument**

```python
def a_function(*fruit):
    print("My Favourite fruit is " + fruit[2])
a_function("Mango","Apple","Grapes")
```

```python
def a_function(**fruit):
    print("My Favourite fruite recepie is " + fruit["recepie"])
a_function(name="Orange",recepie="Orange Juice")
```

```python
x = 50

def show():
    print(x)

print(x)
```
