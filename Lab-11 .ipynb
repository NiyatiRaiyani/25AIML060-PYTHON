**List Comprehension**

**Dict and set Comprehension**

```python
sentence = "Python is easy to learn"

word_lengths = [len(word) for word in sentence.split()]

print(word_lengths)
```

```python
words = ["apple", "banana", "cat", "dog", "apple", "cat"]

unique_lengths = {len(word) for word in words}

print(unique_lengths)
```

```python
sentence = "I"

word_size = {len(word) for word in sentence.split()}

print(word_size)
```

**Nested List Comprehension**

```python
matrix = [[i * j for j in range(1, 4)] for i in range(1, 4)]

print(matrix)
```

```python
students = [
    ["Niyati", 20],
    ["Priya", 21],
    ["Krina", 19]
]

result = [[name, data] for name, data in students]

print(result)
```

```python
names = ["Niyati", "Priya", "Neha", "Riya", "Meera",
         "Karan", "Deep", "Krina", "Het", "Pooja"]

result = [name for name in names if 'e' in name.lower()]

print(result)
```

```python
result = [(name, name.lower().count('e')) for name in names]

print(result)
```
