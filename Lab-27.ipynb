### Part A — Create and Inspect

### Part B — Missing Data Cleaning

4. Find the number of missing values in each column.
5. Fill missing Age values with the mean age.
6. Fill missing Salary values with the median salary.
7. Remove rows where City is missing.
8. Display the cleaned DataFrame.

```python
# 4. Find the number of missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

# 5. Fill missing Age values with mean age
df["Age"] = df["Age"].fillna(df["Age"].mean())

# 6. Fill missing Salary values with median salary
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# 7. Remove rows where City is missing
df = df.dropna(subset=["City"])

# 8. Display the cleaned DataFrame
print("\nCleaned DataFrame:")
print(df)
```

### Part C — Employee Analysis

9. Display only the Name, Department, and Salary columns.
10. Display employees whose salary is greater than ₹50,000.
11. Add a Bonus column equal to 8% of Salary.
12. Sort employees by Salary in descending order.
13. Find the employee with the highest salary.
14. Calculate the average salary department-wise.
15. Find the average experience department-wise.

```python
# 9. Display only Name, Department, and Salary
print("Name, Department and Salary:")
print(df[["Name", "Department", "Salary"]])

# 10. Employees whose salary is greater than ₹50,000
print("\nEmployees with salary greater than ₹50,000:")
print(df[df["Salary"] > 50000])

# 11. Add Bonus column equal to 8% of Salary
df["Bonus"] = df["Salary"] * 0.08

print("\nDataFrame with Bonus:")
print(df[["Name", "Salary", "Bonus"]])

# 12. Sort employees by Salary in descending order
df_sorted = df.sort_values(by="Salary", ascending=False)

print("\nEmployees sorted by Salary (descending):")
print(df_sorted)

# 13. Find the employee with the highest salary
highest_salary_employee = df.loc[df["Salary"].idxmax()]

print("\nEmployee with the highest salary:")
print(highest_salary_employee)

# 14. Calculate average salary department-wise
avg_salary_department = df.groupby("Department")["Salary"].mean()

print("\nAverage salary department-wise:")
print(avg_salary_department)

# 15. Find average experience department-wise
avg_experience_department = df.groupby("Department")["Experience"].mean()

print("\nAverage experience department-wise:")
print(avg_experience_department)
```
