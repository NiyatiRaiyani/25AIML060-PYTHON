**Managing Columns and Rows**

144. Delete a column using del keyword
• Remove unwanted columns using del df['column'].
• Observe effect on original DataFrame.
• Compare with drop() for more flexibility.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Niyati', 'Mahi', 'Meshva'],
    'Age': [21, 22, 20],
    'Marks': [85, 90, 78]
})

print("Original DataFrame:")
print(df)

del df['Age']

print("\nDataFrame after deleting Age:")
print(df)
```

145. Drop rows or columns using drop() with axis argument
• Remove rows using label list.
• Drop columns by setting axis=1.
• Understand that drop() returns a new DataFrame.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Niyati', 'Priya', 'Nency', 'Neha'],
    'Marks': [85, 90, 78, 88],
    'Grade': ['B', 'A', 'C', 'B']
}, index=['S1', 'S2', 'S3', 'S4'])

print("Original DataFrame:")
print(df)

# Drop rows
new_df1 = df.drop(['S2', 'S4'], axis=0)

print("\nAfter dropping rows:")
print(new_df1)

# Drop column
new_df2 = df.drop('Grade', axis=1)

print("\nAfter dropping Grade column:")
print(new_df2)

print("\nInplace dataframe after dropping Grade column:")
df.drop('Grade', axis=1, inplace=True)
print(df)
```

**Operations and Arithmetic**

146. Add two Series with mismatched indexes
• Perform arithmetic with index alignment.
• Understand how NaN appears for unmatched labels.
• Handle missing data post-operation.

```python
import pandas as pd

s1 = pd.Series(
    [10, 20, 30],
    index=['A', 'B', 'C']
)

s2 = pd.Series(
    [5, 15, 25],
    index=['B', 'C', 'D']
)

result = s1 + s2

print("Series 1:")
print(s1)

print("\nSeries 2:")
print(s2)

print("\nResult:")
print(result)

# Handle NaN values
print("\nAfter replacing NaN with 0:")
print(result.fillna(0))
```
