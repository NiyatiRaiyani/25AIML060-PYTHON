**1.Create and Write to a file**

```python
with open("my_notes.txt","w") as file:
    file.write("Today I learned about file handling in Python.\n")
    file.write("It is very useful for reading and writng files.\n")
```

**2.Read a File and Print with Line Numbers**

```python
with open("my_notes.txt", "r") as file:
    lines = file.readlines()
    for i, line in enumerate(lines, start=1):
        print(f"Line {i}: {line.strip()}")
```

**3.Append to a File**

```python
with open("my_notes.txt", "a") as file:
    file.write("I will practice more to become confident.\n")
```

**4.Count Words in a File**

```python
with open("my_notes.txt", "r") as file:
    text = file.read()
    words = text.split()
    print("Total words:", len(words))
```

**5.Copy contents to another file**

```python
with open("my_notes.txt", "r") as source:
    content = source.read()

with open("backup_notes.txt", "w") as target:
    target.write(content)
```

**6.Count Lines, Words, Characters**

```python
with open("my_notes.txt", "r") as file:
    lines = file.readlines()

num_lines = len(lines)
num_words = sum(len(line.split()) for line in lines)
num_chars = sum(len(line) for line in lines)

print("Lines:", num_lines)
print("Words:", num_words)
print("Characters:", num_chars)
```

**7.Remove Blank Lines**

```python
with open("my_notes.txt", "r") as infile, open("cleaned_notes.txt", "w") as outfile:
    for line in infile:
        if line.strip():  # Skip empty lines
            outfile.write(line)
```

**salary = [20000,30000,45000,35000,6000,21000,40000]
1.filter the salaries greater than 30000 using lambda function
2.show top 5% salaries from the list using lamda function
3.square roots of salaries from the list , rounded to integer using lamda function**

**1. Salaries greater than 30000**

```python
salary = [20000, 30000, 45000, 35000, 6000, 21000, 40000]

result = list(filter(lambda x: x > 30000, salary))
print(result)
```

**2. Top 5% salary**

```python
salary = [20000, 30000, 45000, 35000, 6000, 21000, 40000]

top_5 = sorted(salary, key=lambda x: x, reverse=True)[:max(1, round(len(salary) * 0.05))]

print(top_5)
```

**3. Square root of salaries, rounded to integer**

```python
import math

salary = [20000, 30000, 45000, 35000, 6000, 21000, 40000]

result = list(map(lambda x: round(math.sqrt(x)), salary))

print(result)
```
