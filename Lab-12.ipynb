**use of lambda function with filter()**

```python
my_list = [1,2,5,6,7,8,9,4]
```

```python
new_list = list(filter(lambda x :(x%2==0),my_list))
```

```python
print(new_list)
```

**use of lambda function with map()**

```python
my_list = [1, 2, 5, 6, 7, 8, 9, 4]
new_list = list(map(lambda x: x * 2, my_list))
print(new_list)
```

```python
names = ["niyati", "priya", "krina"]
result = map(lambda name: name.upper(), names)
print(list(result))
```

**1. names = ["niyati", "riya", "krina", "mahi"]
Filter the names starting with R**

```python
names = ["niyati", "riya", "krina", "mahi"]
result = list(filter(lambda name: name.startswith("r"), names))
print(result)
```

**try Except block**

```python
def try_float(x):
    try:
        return float(x)
    except(TypeError , ValueError):
        return x
```

```python
try_float((1,2))
```

**Else Block**

```python
def divide(x,y):
    try:
        result = x/y
    except ZeroDivisionError:
        print("Cannot divide a number by zero")
    else:
        print("Dividion Successfull Answer is :" ,result)
        
```

```python
divide(5,2)
```

```python
divide(5,0)
```

**Finally Block**

```python
def divide(x,y):
    try:
        result = x/y
    except ZeroDivisionError:
        print("Cannot divide a number by zero")
    else:
        print("Dividion Successfull Answer is :" ,result)
    finally:
        print("Division Operator Complete")
```

```python
divide(5,2)
```

```python
divide(50,0)
```
