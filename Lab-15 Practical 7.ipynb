**Splitting Array**

```python
import numpy as np
a = np.arange(20)
print('First array')
print(a)
print('\n')
print('Splitting the array in 2 equal size array')
b = np.split(a,2)
print(b)
print('\n')
```

```python
a=np.arange(10)
#splitting the array indicated first at 3, second 5, 3rd on 9
print('splitting the array indicated first at 3, second 5 and remaining in 3rd')
c=np.split(a,[3,5,9])
print(c)
print('\n')
```

84. Write a program to apply scalar arithmetic operations on a NumPy array
• Perform addition, subtraction, multiplication, and division with scalars.
• Understand how operations are broadcast across the array.
• Predict element-wise results for each operation.

```python
import numpy as np

a = np.array([10, 20, 30, 40])

print("Original Array:", a)

print("Addition:", a + 5)
print("Subtraction:", a - 5)
print("Multiplication:", a * 5)
print("Division:", a / 5)
```

85. Write a program to perform element-wise arithmetic between two arrays
• Use +, -, *, / operators between same-shaped arrays.
• Understand and apply broadcasting rules.
• Visualize how array shapes influence operations.

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Array A:", a)
print("Array B:", b)

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

87. Write a program to apply NumPy’s sin(), sqrt(), and log() on arrays
• Use ufuncs for element-wise mathematical transformations.
• Understand the input-output behavior of sin(), log(), and sqrt().
• Compare original and transformed arrays.

```python
import numpy as np

a = np.array([1, 4, 9, 16])

print("Original Array:", a)

print("Square Root:", np.sqrt(a))
print("Sine:", np.sin(a))
print("Log:", np.log(a))
```

88. Write a program to multiply an array with a ufunc result (e.g., array *
sin(array))
• Combine arrays and ufunc results for advanced element-wise operations.
• Chain ufuncs in expressions.
• Visualize how mathematical functions modify array data.

```python
import numpy as np

a = np.array([1, 2, 3, 4])

print("Original Array:", a)

result = a * np.sin(a)

print("Sin of Array:", np.sin(a))
print("Array * Sin(Array):", result)
```
