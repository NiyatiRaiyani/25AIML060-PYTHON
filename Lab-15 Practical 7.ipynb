**Splitting Array**

```python
import numpy as np
a = np.arange(20)
print('First array')
print(a)
print('\n')
print('Splitting the array in 2 equal size array')
b = np.split(a,2)
print(b)
print('\n')
```

```python
import numpy as np
a = np.arange(10)
print('Splitting the array: First at 3, second from 3 to 5, third from 5 to 9, remaining in fourth')
c = np.split(a, [3, 5, 9])
print(c)
```
