**Handling Missing Data (NaNs)**

127. Write a program to filter valid and missing values using Boolean masks
• Use notnull() to select valid entries.
• Use isnull() to isolate missing entries.
• Understand how these tools are useful for data cleaning.

```python
import pandas as pd
import numpy as np

s = pd.Series([10, np.nan, 30, np.nan, 50])

valid = s.notnull()
missing = s.isnull()

print("Valid values:")
print(s[valid])

print("\nMissing values:")
print(s[missing])
```

**Combining and Aligning Series**

128. Write a program to add two Series and observe label alignment
• Perform addition on two Series with overlapping and distinct labels.
• Interpret how missing label combinations result in NaN.
• Understand the label-driven alignment in Series operations.

```python
import pandas as pd

s1 = pd.Series([100, 200, 300], index=['A', 'B', 'C'])
s2 = pd.Series([50, 150, 250], index=['B', 'C', 'D'])

result = s1 + s2

print("Series 1:")
print(s1)

print("\nSeries 2:")
print(s2)

print("\nAddition:")
print(result)
```
