Defining a Dataframe

```python
import pandas as pd
data = {'colors': ['red', 'blue', 'green', 'yellow', 'black'],
        'items': ['ball', 'pen', 'pencil', 'eraser', 'sharpener'],
        'price': [2000, 1000, 3000, 5000, 4000]}
frame = pd.DataFrame(data)
frame
```

```python
frame['new'] = 12
frame
```

```python
frame['new'] = [12,14,85,96,78]
frame
```

```python
import numpy as np
ser = pd.Series(np.arange(5))
ser
```

```python
frame ['new'] = ser
frame
```

Assigning Values

```python
frame['price'][2]=1000
frame
```

```python
frame.loc[2,'price']=2500
frame
```

Membership of a Value

```python
frame.isin([1.0,'pen'])
```

```python
frame[frame.isin([1.0,'pen'])]
```

Deleting a Column

```python
del frame['new']
frame
```

 Filtering

```python
frame[frame < 1.2]
frame
```
