Step 1 – Import Pandas

```python
import numpy as np
import pandas as pd
```

Step 2-Load the dataset

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\hppc\Downloads\amazon - amazon.csv")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
df.head(10)
```

Step-3 Understand the dataset

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()
```

Step-4 Check missing values

```python
missing_values = df.isnull().sum()
missing_values.sort_values(ascending=False)
```

Step-5 Clean numerical columns

```python
# Clean numerical columns

df['discounted_price'] = (
    df['discounted_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)

df['actual_price'] = (
    df['actual_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)

df['discount_percentage'] = (
    df['discount_percentage']
    .astype(str)
    .str.replace('%', '', regex=False)
    .astype(float)
)

df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

df["rating_count"] = (
    df["rating_count"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .replace("nan", np.nan)
)
df["rating_count"] = pd.to_numeric(df["rating_count"], errors="coerce")

df[["discounted_price", "actual_price", "discount_percentage",
    "rating", "rating_count"]].head()
```

Step-6 create a discount amount column

```python
df["discount_amount"] = (
    df["actual_price"] - df["discounted_price"]
)

df[["product_name", "actual_price", "discounted_price", "discount_amount"]].head()
```

Step 7 - use Numpy for statiscal calculations

```python
ratings = df["rating"].dropna().to_numpy()

print(ratings)
print("Average rating:", np.mean(ratings))
print("Minimum rating:", np.min(ratings))
print("Maximum rating:", np.max(ratings))
print("Standard deviation:", np.std(ratings))
```

Step-8 Find highly rated products using Numpy

```python
df["rating_status"] = np.where(
    df["rating"] >= 4.0,
    "Highly Rated",
    "Below 4.0"
)

df[["product_name", "rating", "rating_status"]].head(10)
```

Step - 9 - Find products with high discounts

```python
high_discount = df[df["discount_percentage"] >= 50]

print("Number of products with 50% or more discount:",
      len(high_discount))

high_discount.head(10)
```

Step-10 Find the top 10most highly rated products

```python
top_rated = (
    df[df["rating_count"] >= 100]
    .sort_values(["rating", "rating_count"], ascending=[False, False])
)

top_rated[
    ["product_name", "rating", "rating_count", "discounted_price"]
].head(10)
```

Step-11 Category Analysis

```python
list(df['category'].head(10))
```

```python
df["main_category"] = df["category"].astype(str).str.split("|").str[0]

df["main_category"].value_counts().head(10)
```

Step-12 Calculate category wise statics

```python
category_summary = (
    df.groupby("main_category")
      .agg(
          product_count=("product_id", "count"),
          average_rating=("rating", "mean"),
          average_discount=("discount_percentage", "mean"),
          average_price=("discounted_price", "mean")
      )
      .sort_values("product_count", ascending=False)
)

category_summary.head(10)
```

Step-13 find the most reviewed products

```python
most_reviewed = df.sort_values(
    "rating_count", ascending=False
)

most_reviewed[
    ["product_name", "rating", "rating_count", "discounted_price"]
].head(10)
```

Step 14 — NumPy: calculate the overall price statistics

```python
prices = df["discounted_price"].dropna().to_numpy()

print("Average discounted price:", np.mean(prices))
print("Median discounted price:", np.median(prices))
print("Minimum discounted price:", np.min(prices))
print("Maximum discounted price:", np.max(prices))
```

Step-15 create a simple final summery

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Products",
        "Average Rating",
        "Average Discount %",
        "Average Discounted Price",
        "Products with >= 50% Discount"
    ],
    "Value": [
        len(df),
        round(df["rating"].mean(), 2),
        round(df["discount_percentage"].mean(), 2),
        round(df["discounted_price"].mean(), 2),
        int((df["discount_percentage"] >= 50).sum())
    ]
})

summary
```
