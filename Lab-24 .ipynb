Step 1 – Import Pandas

```python
import numpy as np
import pandas as pd
```

Step 2-Load the dataset

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\hppc\Downloads\amazon - amazon.csv")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

Step-3 Understand the dataset

```python
print("Column name:" ,df.columns.tolist())
print("\nData types:" ,df.dtypes)
print("\nDataset information :" ,df.info())
```

Step-4 Check missing values

```python
missing_values = df.isnull().sum()
print("Missing Values in Each Column:")
print(missing_values.sort_values())

print("\nTotal Missing Values:")
print(missing_values.sum())
```

Step-5 Clean numerical columns

```python
numeric_columns = df.select_dtypes(include=['number']).columns

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors='coerce')

print("Numerical columns cleaned successfully.")
print(df[numeric_columns].head())
```

```python
# Clean numerical columns

df['discounted_price'] = (
    df['discounted_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)

df['actual_price'] = (
    df['actual_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)

df['discount_percentage'] = (
    df['discount_percentage']
    .astype(str)
    .str.replace('%', '', regex=False)
    .astype(float)
)

df['rating_count'] = (
    df['rating_count']
    .astype(str)
    .str.replace(',', '', regex=False)
    .astype(float)
)

# Display cleaned numerical columns
print(df[['product_name',
          'discounted_price',
          'actual_price',
          'discount_percentage',
          'rating_count']].head().to_string(index=False))
```

Step-6 create a discount amount column

```python
# Calculate the discount amount for each product
df['discount_amount'] = df['actual_price'] - df['discounted_price']

# Display the required columns
print(df[['product_name', 'actual_price',
          'discounted_price', 'discount_amount']].head().to_string(index=False))
```

Step 7 - use Numpy for statiscal calculations

```python
import numpy as np

print("Mean Discounted Price:", np.mean(df['discounted_price']))
print("Median Discounted Price:", np.median(df['discounted_price']))
print("Standard Deviation:", np.std(df['discounted_price']))
print("Minimum Discounted Price:", np.min(df['discounted_price']))
print("Maximum Discounted Price:", np.max(df['discounted_price']))
```

Step-8 Find highly rated products using Numpy

```python
import numpy as np

df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

highly_rated = df[np.greater_equal(df['rating'], 4.5)]

low_rated = df[np.less(df['rating'], 4)]

print("Highly Rated Products (Rating >= 4.5)")
print(highly_rated[['product_name', 'rating']].to_string(index=False))

print("\nProducts Rated Below 4")
print(low_rated[['product_name', 'rating']].to_string(index=False))
```

```python
import numpy as np
df["rating_status"] = np.where(
    df["rating"] >= 4.0,
    "Highly Rated",
    "Below 4.0"
)

print(df[["product_name", "rating", "rating_status"]].head())
```

Step - 9 - Find products with high discounts

```python
high_discount = df[df["discount_percentage"] >= 50]

print(high_discount[
    ["product_name", "actual_price", "discounted_price", "discount_percentage"]
].head(10).to_string(index=False))
```

Step-10 Find the top 10most highly rated products

```python
top_10_rated = df.sort_values("rating", ascending=False).head(10)

result = top_10_rated[["product_name", "rating"]]

print(result.to_string(index=False, justify="left"))
```

Step-11 Category Analysis

```python
category_analysis = df["category"].value_counts()

print(category_analysis.to_string())
```

Step-12 Calculate category wise statics

```python
category_stats = df.groupby("category").agg({
    "rating": "mean",
    "actual_price": "mean",
    "discounted_price": "mean",
    "discount_percentage": "mean"
}).reset_index()

print(category_stats.to_string(index=False))
```
