# My Personal Library Manager
### Python Practical — Professional Jupyter Notebook

**Aim:** To develop a menu-driven Python application to add, display, search, delete, and track books, and save/load book records using file handling.

**Concepts used:** Lists, dictionaries, functions, loops, conditionals, `map()`, `filter()`, `lambda`, sorting, exception handling, and file handling.


## Step 1 — Import Required Module
The `os` module helps check whether the data file exists.

```python
import os

DATA_FILE = "books_data.txt"
print("Required module imported successfully.")
```

## Step 2 — Create the Library List
Each book is stored as a dictionary inside the list.

```python
library = []

print("Library initialized.")
print("Total books:", len(library))
```

## Step 3 — Add a Book
The function accepts the title, author, genre, year, and reading status. Status defaults to `unread`.

```python
def add_book(title, author, genre, year, status="unread"):
    try:
        year = int(year)
        status = status.strip().lower()

        if not title.strip() or not author.strip() or not genre.strip():
            print("Title, author, and genre cannot be empty.")
            return

        if status not in ("read", "unread"):
            print("Status must be 'read' or 'unread'.")
            return

        book = {
            "title": title.strip(),
            "author": author.strip(),
            "genre": genre.strip(),
            "year": year,
            "status": status
        }

        library.append(book)
        print("Book added successfully:", book["title"])

    except ValueError:
        print("Year must be a valid number.")
```

### Step 3.1 — Test Add Book

```python
add_book("Atomic Habits", "James Clear", "Self-Help", 2018, "read")
add_book("The Alchemist", "Paulo Coelho", "Fiction", 1988)
add_book("Sapiens", "Yuval Noah Harari", "History", 2011, "read")
add_book("Deep Work", "Cal Newport", "Productivity", 2016, "unread")
add_book("Becoming", "Michelle Obama", "Biography", 2018, "read")
```

## Step 4 — Display All Books and Sort by Year
The `sorted()` function and `lambda` sort books by publication year without changing the original list.

```python
def list_books(sort_by_year=False):
    if not library:
        print("Library is empty.")
        return

    books_to_show = sorted(library, key=lambda book: book["year"]) if sort_by_year else library

    for number, book in enumerate(books_to_show, start=1):
        print(
            f"{number}. {book['title']} by {book['author']} "
            f"[{book['genre']}, {book['year']}] - {book['status'].capitalize()}"
        )

print("Books in library:")
list_books()

print("\nBooks sorted by year:")
list_books(sort_by_year=True)
```

## Step 5 — Search Books by Author
The search is case-insensitive and accepts a full or partial author name. A list comprehension returns matching books.

```python
def search_by_author(author_name):
    search_text = author_name.strip().lower()
    return [
        book for book in library
        if search_text in book["author"].lower()
    ]

matches = search_by_author("james")
if matches:
    for book in matches:
        print(book["title"], "-", book["author"])
else:
    print("No matching author found.")
```

## Step 6 — Delete a Book by Title
The function removes all entries whose titles match the requested title, ignoring uppercase/lowercase differences.

```python
def delete_book(title):
    global library
    search_title = title.strip().lower()
    old_count = len(library)

    library = [
        book for book in library
        if book["title"].lower() != search_title
    ]

    if len(library) < old_count:
        print("Book deleted successfully.")
    else:
        print("Book not found.")

# Example: delete_book("Deep Work")
```

## Step 7 — Reading Summary
Count read and unread books and display their percentages.

```python
def reading_summary():
    if not library:
        print("Library is empty.")
        return

    total = len(library)
    read_count = sum(1 for book in library if book["status"] == "read")
    unread_count = total - read_count

    print("Total books:", total)
    print(f"Read: {read_count} ({read_count / total * 100:.2f}%)")
    print(f"Unread: {unread_count} ({unread_count / total * 100:.2f}%)")

reading_summary()
```

## Step 8 — Display Uppercase Book Titles
`map()` applies `str.upper` to every title without changing the original book records.

```python
def uppercase_titles():
    titles = list(map(lambda book: book["title"].upper(), library))

    print("Book Titles (UPPERCASE):")
    for title in titles:
        print("-", title)

uppercase_titles()
```

## Step 9 — Save Books to a File
Each book is saved as one comma-separated line in `books_data.txt`. This version uses CSV handling so fields containing commas are handled more safely.

```python
import csv

def save_to_file(filename=DATA_FILE):
    try:
        with open(filename, "w", newline="", encoding="utf-8") as file:
            writer = csv.DictWriter(
                file,
                fieldnames=["title", "author", "genre", "year", "status"]
            )
            writer.writeheader()
            writer.writerows(library)

        print(f"Saved {len(library)} books to {filename}.")
    except OSError as error:
        print("Error saving file:", error)

save_to_file()
```

## Step 10 — Load Books from a File
Check whether the file exists, then read valid records. Invalid records are skipped rather than stopping the whole loading process.

```python
def load_from_file(filename=DATA_FILE):
    global library

    if not os.path.exists(filename):
        print("File not found. Starting with the current library unchanged.")
        return

    loaded_books = []
    try:
        with open(filename, "r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            required_fields = {"title", "author", "genre", "year", "status"}
            if not required_fields.issubset(set(reader.fieldnames or [])):
                print("File format is incorrect.")
                return

            for row in reader:
                try:
                    title = row["title"].strip()
                    author = row["author"].strip()
                    genre = row["genre"].strip()
                    year = int(row["year"])
                    status = row["status"].strip().lower()

                    if title and author and genre and status in ("read", "unread"):
                        loaded_books.append({
                            "title": title,
                            "author": author,
                            "genre": genre,
                            "year": year,
                            "status": status
                        })
                except (ValueError, AttributeError, TypeError):
                    print("Skipped one invalid record.")

        library = loaded_books
        print(f"Loaded {len(library)} books from {filename}.")
    except OSError as error:
        print("Error reading file:", error)

# To test loading, run this after saving:
# load_from_file()
# list_books()
```

## Step 11 — Create a Menu-Driven Program
The menu repeats until the user chooses Exit. Invalid choices and invalid years are handled.

```python
def menu():
    while True:
        print("\n===== PERSONAL LIBRARY MANAGER =====")
        print("1. Add Book")
        print("2. List All Books")
        print("3. List Books Sorted by Year")
        print("4. Search by Author")
        print("5. Delete Book by Title")
        print("6. Show Reading Summary")
        print("7. Show Uppercase Titles")
        print("8. Save Library")
        print("9. Load Library")
        print("10. Exit")

        choice = input("Enter your choice (1-10): ").strip()

        if choice == "1":
            title = input("Enter title: ")
            author = input("Enter author: ")
            genre = input("Enter genre: ")
            year = input("Enter publication year: ")
            status = input("Enter status (read/unread; default unread): ").strip()
            add_book(title, author, genre, year, status or "unread")

        elif choice == "2":
            list_books()

        elif choice == "3":
            list_books(sort_by_year=True)

        elif choice == "4":
            author = input("Enter author name or part of name: ")
            results = search_by_author(author)
            if results:
                for book in results:
                    print(f"{book['title']} by {book['author']} ({book['year']})")
            else:
                print("No matching books found.")

        elif choice == "5":
            delete_book(input("Enter title to delete: "))

        elif choice == "6":
            reading_summary()

        elif choice == "7":
            uppercase_titles()

        elif choice == "8":
            save_to_file()

        elif choice == "9":
            load_from_file()

        elif choice == "10":
            print("Thank you for using Personal Library Manager.")
            break

        else:
            print("Invalid option. Please enter a number from 1 to 10.")
```

## Step 12 — Run the Application
**Important for Jupyter:** run `menu()` in the next cell when you want to interact with the program. The menu waits for input until option 10 is selected.

```python
if __name__ == "__main__":
    menu()
```
