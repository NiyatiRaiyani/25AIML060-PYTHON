**Exrecise Calulator**

```python
first = input("Enter your first number : ")
operator = input("enter operator(+,-,*,/,%) : ")
second = input("Enter your second number : ")

if operator == "+":
    sum = int(first) + int(second)
    print("The sum is: " + str(sum))    

elif operator == "-":
    difference = int(first) - int(second)
    print("The difference is: " + str(difference))

elif operator == "*":
    product = int(first) * int(second)
    print("The product is: " + str(product))

elif operator == "/":
    quotient = int(first) / int(second)
    print("The quotient is: " + str(quotient))

elif operator == "%":
    remainder = int(first) % int(second)    
    print("The remainder is: " + str(remainder))

else:
    print("Invalid operator")
```

**break and continue**

```python
for n in range(1,51):
    if n == 21:
        break
    if n % 3 == 0:
        print(n)
print("Out of loop")
```

```python
for n in range(1,51):
    if n == 21:
        continue
    if n % 3 == 0:
        print(n)
print("Out of loop")
```
