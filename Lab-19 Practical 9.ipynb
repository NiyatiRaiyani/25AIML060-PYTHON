```python
A = np.array([[1, 2, 3],[4, 5, 6]])

b = np.array([10, 20, 30])

result = A + b

print("A shape:", A.shape)
print("b shape:", b.shape)
print("Result:" , result)
print("Result shape:", result.shape)
```
