**Broadcasting Basics**

113. Write a program to apply broadcasting between a 2D and 1D array
• Use broadcasting to apply operations on arrays of different shapes.
• Understand NumPy's broadcasting rules (same or size-1 dimensions).
• Predict result shapes after operations like A + b.

```python
import numpy as np
```

```python
A = np.array([[1, 2, 3],[4, 5, 6]])

b = np.array([10, 20, 30])

result = A + b

print("A shape:", A.shape)
print("b shape:", b.shape)
print("Result:" )
print(result)
print("Result shape:", result.shape)
```

**Advanced Broadcasting with Higher Dimensions**

115. Write a program to multiply two 3D arrays using complex broadcasting
• Reshape arrays to (3,1,2) and (3,2,1) and apply broadcasting.
• Understand compatibility conditions for higher-dimensional arrays.
• Visualize resulting 3D array after broadcasting multiplication.

```python
A = np.arange(6).reshape(3, 1, 2)
B = np.arange(6).reshape(3, 2, 1)

result = A * B

print("A shape:", A.shape)
print("B shape:", B.shape)
print("Result shape:", result)
print("Result:")
print(result)
```

116. Write a program to verify broadcasting compatibility rules for mismatched
shapes
• Apply Rule 1: Insert 1 in missing dimensions.
• Apply Rule 2: Replicate along dimensions of size 1.
• Test and resolve shape mismatch errors in real use cases.

```python
A = np.ones((2, 3))
B = np.ones((3,))

try:
    result = A + B
    print("Compatible")
    print("Result shape:", result.shape)
except ValueError:
    print("Not compatible")

# Mismatched example
C = np.ones((2, 3))
D = np.ones((2,))

try:
    result = C + D
    print("Compatible")
    print("Result shape:", result.shape)
except ValueError:
    print("Not compatible: Shape mismatch")
```

**Creating Series from Various Sources**

117. Write a program to create a Series from a list with and without custom index
• Create a basic Series using a list of numeric values.
• Apply custom labels as index instead of default integer index.
• Interpret the string representation of a Series.

```python
import pandas as pd

data = [10, 20, 30, 40]

# Without custom index
s1 = pd.Series(data)

# With custom index
s2 = pd.Series(data, index=['A', 'B', 'C', 'D'])

print("Series without custom index:")
print(s1)

print("\nSeries with custom index:")
print(s2)
```

118. Write a program to create a Series from a NumPy array and verify data linkage
• Convert a NumPy array into a pandas Series.
• Observe dynamic linking between original array and Series.
• Understand mutability and reference behavior.

```python
import numpy as np
import pandas as pd

arr = np.array([10, 20, 30, 40])

s = pd.Series(arr, copy=False)

print("Original NumPy array:")
print(arr)

print("\nOriginal Series:")
print(s)

arr[0] = 100

print("\nAfter changing NumPy array:")
print(arr)

print("\nSeries after changing NumPy array:")
print(s)
```

119. Write a program to create a Series from a dictionary with custom index
• Use dictionary key-value pairs to form Series index and data.
• Handle missing keys using an index list (introducing NaN).
• Understand auto-alignment of labels and handling of missing data.

```python
import pandas as pd

data = {
    'Laptop': 50000,
    'Desktop': 20000,
    'Tablet': 30000,
    'Smartphone': 40000
}

s = pd.Series(data, index=['Laptop', 'Desktop', 'Tablet', 'Smartphone'])  

print(s)
```

```python
import pandas as pd
marks = np.array([75, 85, 90, 95, 80])
series = pd.Series(marks, index=['Math', 'Science', 'English', 'History', 'Art'])
print(series)
```

```python
import pandas as pd
marks = pd.Series([75, 85, 90, 95, 80],
                  index=['Niyati', 'Aarav', 'Riya', 'Kabir', 'Anaya'])

print("Original Series:")
print(marks)

print("\nMarks of Niyati:", marks['Niyati'])
print("\nMarks of 3rd student:", marks.iloc[2])
print("\nMarks of First three students:", marks.iloc[:3])
print("\nMarks of Aarav and Kabir :",marks[['Aarav', 'Kabir']])
```

```python
import pandas as pd
stock = pd.Series([50000, 20000, 30000, 40000],
                  index=['Laptop', 'Desktop', 'Tablet', 'Smartphone'])

print("Original stock")
print(stock)

stock['Laptop'] = 55000
stock.iloc[1] = 25000

stock[['Tablet', 'Smartphone']] = [35000, 45000]

stock[['Tablet', 'Smartphone']] = stock[['Tablet', 'Smartphone']] + 1000

print("\nUpdated stock")
print(stock)
```
